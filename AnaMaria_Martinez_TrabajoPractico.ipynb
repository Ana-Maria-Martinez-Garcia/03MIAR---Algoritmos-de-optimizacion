Descripción del problema:

Se precisa coordinar el doblaje de una película. Los actores del doblaje deben coincidir en las tomas en las que sus personajes aparecen juntos en las diferentes tomas. Los actores de doblaje cobran todos la misma cantidad por cada día que deben desplazarse hasta el estudio de grabación independientemente del número de tomas que se graben. No es posible grabar más de 6 tomas por día. El objetivo es planificar las sesiones por día de manera que el gasto por los servicios de los actores de doblaje sea el menor posible. Los datos son:
  - Número de actores: 10
  - Número de tomas: 30
  - Actores/Tomas: https://bit.ly/36D8IuK

Un 1 indica que el actor participa en la toma y un 0 que no.






                                        

#Modelo
- ¿Como represento el espacio de soluciones?
- ¿Cual es la función objetivo?
- ¿Como implemento las restricciones?

**¿Cómo represento el espacio de soluciones?**

La solución del problema consiste en una lista de listas, donde cada sublista representa las tomas que se deben realizar en ese día.

En concreto, la lista constará de 5 sublistas (5 días), y cada una de estas estará formada por 6 elementos (las 6 tomas de ese día).

**¿Cuál es la función objetivo?**

La función objetivo viene determinada por el coste total acumulado a lo largo de todos los días de grabación. El coste de cada día depende del número de actores distintos que participan en las tomas de ese día.

**¿Cómo implemento las restricciones?**

Las restricciones del problema son las siguientes:

1. Los actores del doblaje deben coincidir en las tomas en las que sus personajes aparecen juntos. Esta restricción queda implementada ya que cada toma se considera indivisible y se asigna íntegramente a un único día. En el código esto se refleja al calcular los actores del día como la unión de actores de sus tomas.

2. No es posible grabar más de 6 tomas por día. En la estrategia voraz cada día se contruye añadiendo tomas hasta que el largo de la lista que define ese día es igual a 6. En la búsqueda tabú, los movimientos consisten en intercambiar tomas entre días, por lo que no se modifica el tamaño de las listas. Puesto que se deben grabar 30 tomas y no pueden realizarse más de 6 por día, la planificación se distribuye en 5 días, quedando cada uno de ellos con 6 tomas asignadas.

3. Cada toma debe asignarse solo a un día. En el algoritmo voraz, cuando una toma se añade a un día se elimina del conjunto de tomas no asignadas, evitando que pueda volver a asignarse. Como se ha comentado en la restricción anterior, en la búsqueda tabú únicamente se llevan a cabo intercambios de tomas ya asignadas, por lo que se conserva el conjunto total de tomas.


#Análisis
- ¿Que complejidad tiene el problema?. Orden de complejidad y Contabilizar el espacio de soluciones

**Contabilización del espacio de soluciones.**

El número de soluciones factibles corresponde al número de formas de dividir un conjunto de n elementos en r grupos de tamaño k, sin importar el orden dentro de cada grupo ni el orden de los grupos. Este valor viene dado por la fórmula combinatoria: (n!)/((k!)^r * r!).

En nuestro problema, n=30 tomas, r=5 días y k=6 tomas por día, por lo que el número de soluciones posibles es: (30!)/ ((6!)^5 * 5!).

Este valor es muy grande, lo que hace inviable aplicar algoritmos de búsqueda exhaustiva.

**Orden de complejidad del algoritmo voraz.**

Este algortimo consta de 2 pasos. Para estudiar la complejidad de cada uno de ellos por separado, y finalmente en su conjunto, se fijan los siguientes parámetros:
- n = número de tomas (30)
- m = número de actores (10)

Paso 1: selección de la primera toma de cada día.
Para inicializar cada día se implementa un bucle que recorre el conjunto de tomas que aún no han sido asignadas (`for t in tomas_no_asignadas:`). Este bucle tiene una complejidad O(n). Sin embargo, como este bucle se repite para cada día (aunque cada vez con menos cantidad de tomas restantes), el coste acumulado del paso 1 queda como O(n^2).

Paso 2: completar el día con las 5 tomas restantes.
Para completar el día se vuelven a recorrer las tomas que aún no han sido asignadas (`for t in tomas_no_asignadas:`) y para cada una de ellas se comprueba cuántos actores nuevos se necesitarían al añadirla (`incremento_actores(dia, t)`). La función incremento_actores tiene un coste O(m).
Como este proceso vuelve a repetirse varias veces sobre el conjunto de tomas restantes (el cuál tendrá cada vez menos elementos), el coste total del paso 2 es O(n^2 * m).

Por tanto, el orden de complejidad del algoritmo voraz es: O(n^2 *m).

**Orden de complejidad de la búsqueda tabú.**

Al igual que en el apartado anterior, vamos a empezar fijando el valor de algunos parámetros:
- n = número de tomas (30)
- m = número de actores (10)
- k = número de tomas por día (6)
- d = número de días (5)
- Tmax = número máximo de iteraciones de la búsqueda tabú (1000)

La búsqueda tabú consta de un bucle principal que se ejecuta hasta Tmax iteraciones y, en cada iteración, explora un vecindario de soluciones generado mediante intercambios de tomas entre días distintos.

En primer lugar, el bucle principal (`for t in range(1, Tmax + 1):`) aporta un coste O(Tmax).

En la exploración del vecindario primero se recorren los pares de días con dos bucle anidados (`for d1 in range(len(S)): for d2 in range(d1 + 1, len(S)):`) y para cada par de días, se recorren las posiciones de las tomas dentro de cada día (`for i1 in range(len(S[d1])): for i2 in range(len(S[d2])):`).

Dado que el número de pares de días posibles es el coeficiente binomial (d 2) y que cada día tiene k tomas (por lo que el número de intercambios por par de días es k^2), el número total de intercambios evaluados en cada iteración es (d 2) * k^2.

La evaluación de cada intercambio (`dif = diferencia_coste_cambio(S, d1, i1, d2, i2)`) tiene un coste O(m).

En consecuencia, el coste por iteración es O((d 2) * k^2 * m) y el coste total del algoritmo queda acotado por O(Tmax * (d 2) * k^2 * m).

En esta instancia concreta, dado que d y k son constantes pequeñas (5 y 6 respectivamente), el crecimiento está dominado por el factor Tmax * m, lo que hace que el algoritmo sea computacionalmente asumible.

**¿Qué técnica utilizo? ¿Por qué?**

Tras analizar el tamaño del espacio de soluciones se descartó la idea de emplear un método exacto, ya que probar todas las combinaciones sería inviable.

Para decidir qué estrategia utilizar, además del análisis teórico del problema, se han revisado ejemplos en la literatura sobre problemas similares de planificación y agrupación.

Finalmente, se decidió utilizar un algoritmo voraz para obtener una primera solución factible, y posteriormente aplicar búsqueda tabú para mejorarla.

El algoritmo voraz permite obtener una primera solución factible y rápida en la que se encuentran días con alto grado de solapamiento de actores. No obstante, tal y como se ha visto en clase, puede quedarse atrapado en óptimos locales.

Para solucionar este problema se empleó búsqueda tabú. Esta técnica acepta movimientos que no son necesariamente mejores, pero que dan la oportunidad de escapar de óptimos locales. Además, gracias al uso de la memoria tabú, se evitan ciclos.

En este problema concreto, la combinación de ambas técnicas permitió mejorar la solución inicial obtenida por el método voraz.

```
import pandas as pd

ruta = "/content/Datos problema doblaje(30 tomas, 10 actores) - Hoja 1.csv"
df = pd.read_csv(ruta, skiprows=1, index_col=0)
df = df.iloc[:30, :10]
df = df.astype(int)
df.columns = [f"Actor{c}" for c in df.columns]

print(df.head(5))
```

```
df.index = df.index.astype(int)
#Creamos un diccionario que asocia a cada toma la lista de actores que participan en ella
tomas = {toma: list(df.columns[df.loc[toma] == 1]) for toma in df.index}
```

```
MAX_TOMAS_DIA = 6

#Devuelve los actores que deben acudir si se graban las tomas de 'dia'
def actores_por_tomas(dia):
  s = set()
  for t in dia:
      s.update(tomas[t])
  return s

#Cuántos actores se encesitan para las tomas de 'dia'
def coste_dia(dia):
  return len(actores_por_tomas(dia))

#Cuántos actores nuevos se necesitan si añado esa toma a 'día'
def incremento_actores(dia, toma_candidata):
  antes = actores_por_tomas(dia)
  despues = antes | set(tomas[toma_candidata])
  return len(despues) - len(antes)

#Implementación del algoritmo voraz
def coordinar_doblaje_voraz():
  tomas_no_asignadas = set(tomas.keys())
  sol = [] #será una lista de días, donde cada día es una lista de 6 tomas

  while tomas_no_asignadas:
      #La primera ronda que vamos a escoger para cada día es la que más actores necesite
      toma_mayor_num_actores = None
      mayor_num_actores = -1

      for t in tomas_no_asignadas:
          num_actores = len(tomas[t])
          if num_actores > mayor_num_actores:
              mayor_num_actores = num_actores
              toma_mayor_num_actores = t
          elif num_actores == mayor_num_actores and t < toma_mayor_num_actores:
              toma_mayor_num_actores = t

      dia = [toma_mayor_num_actores]
      tomas_no_asignadas.remove(toma_mayor_num_actores)

      #Elegida una toma, completamos el resto del día con 5 tomas que tengan el menor num de actores nuevos posible
      while len(dia) < MAX_TOMAS_DIA and tomas_no_asignadas:
          actores_actuales = actores_por_tomas(dia)

          mejor_toma = None
          menor_incremento = None
          mayor_solapamiento = None

          for t in tomas_no_asignadas:
              inc = incremento_actores(dia, t)
              solapamiento = len(actores_actuales & set(tomas[t]))

              if mejor_toma is None:
                  mejor_toma = t
                  menor_incremento = inc
                  mayor_solapamiento = solapamiento
              else:
                  if inc < menor_incremento:
                      mejor_toma = t
                      menor_incremento = inc
                      mayor_solapamiento = solapamiento
                  elif inc == menor_incremento and solapamiento > mayor_solapamiento:
                      mejor_toma = t
                      menor_incremento = inc
                      mayor_solapamiento = solapamiento
                  elif inc == menor_incremento and solapamiento == mayor_solapamiento and t < mejor_toma: #por reproducibilidad simplemente
                      mejor_toma = t

          dia.append(mejor_toma)
          tomas_no_asignadas.remove(mejor_toma)

      sol.append(dia)

  return sol

sol_voraz = coordinar_doblaje_voraz()

print("Coste total:", sum(coste_dia(d) for d in sol_voraz))

for i, d in enumerate(sol_voraz, 1):
  print(f"Día {i}: tomas={d} coste_dia={coste_dia(d)}")
```

```
#Calcula el coste total de una solución
def coste_solucion(sol):
  coste = 0
  for dia in sol:
    coste += coste_dia(dia)
  return coste

#Intercambia dos tomas entre dos días distintos.
def intercambiar(sol, d1, i1, d2, i2):
  import copy
  nueva = [dia[:] for dia in sol]
  nueva[d1][i1], nueva[d2][i2] = nueva[d2][i2], nueva[d1][i1]
  return nueva

#Calcula cuánto cambia el coste al cambiar dos tomas de días.
def diferencia_coste_cambio(sol, d1, i1, d2, i2):
  dia1 = sol[d1]
  dia2 = sol[d2]
  coste_antes = coste_dia(dia1) + coste_dia(dia2)

  nueva_sol = intercambiar(sol, d1, i1, d2, i2)
  coste_despues = coste_dia(nueva_sol[d1]) + coste_dia(nueva_sol[d2])

  return coste_despues - coste_antes

#Búsqueda tabú
def tabu_search_expiracion(sol_inicial, Tmax=1000, TabuTenure=12):

    # -- Paso1: INICIALIZACIÓN --
    S = [dia[:] for dia in sol_inicial]
    Z = coste_solucion(S) #valor objetivo

    MejorSol = [dia[:] for dia in S]
    MejorZ = Z

    #Lista tabú por expiración
    tabu_hasta = {}

    ## -- Paso2: MOVIMIENTOS --
    for t in range(1, Tmax + 1):
        mejor_mov = None
        mejor_dif = None
        mejor_mov_id = None
        mejor_Z_nueva = None

        #Explorar vecindario
        for d1 in range(len(S)):
            for d2 in range(d1 + 1, len(S)):
                for i1 in range(len(S[d1])):
                    for i2 in range(len(S[d2])):
                        tomaA = S[d1][i1]
                        tomaB = S[d2][i2]
                        #Para no hacer cambios entre las mismas tomas a la inversa, ya que es lo mismo
                        mov_id = (tomaA, tomaB) if tomaA < tomaB else (tomaB, tomaA)

                        dif = diferencia_coste_cambio(S, d1, i1, d2, i2)
                        Z_nueva = Z + dif

                        #Comprobamos si en la iteración actual sigue siendo tabú
                        es_tabu = tabu_hasta.get(mov_id, 0) >= t

                        #Si es tabú y encima no mejora la mejor sol hasta el momento, se descarta
                        if es_tabu and Z_nueva >= MejorZ:
                            continue

                        # -- Paso3: EVALUACION --
                        #Elegir el mejor candidato permitido
                        if (mejor_dif is None) or (dif < mejor_dif) or (dif == mejor_dif and mov_id < mejor_mov_id):
                            mejor_dif = dif
                            mejor_mov = (d1, i1, d2, i2)
                            mejor_mov_id = mov_id
                            mejor_Z_nueva = Z_nueva

        # --Paso4: ACTUALIZACIÓN --
        if mejor_mov is None:
            break

        #Aplicar el movimiento elegido
        S = intercambiar(S, *mejor_mov)
        Z += mejor_dif

        #Actualizar mejor global
        if Z < MejorZ:
            MejorZ = Z
            MejorSol = [dia[:] for dia in S]

        #Marcar como tabú hasta t + tenure
        tabu_hasta[mejor_mov_id] = t + TabuTenure

    return MejorSol, MejorZ

sol, z = tabu_search_expiracion(sol_voraz)
print("La solución es:", sol, "y el coste:", z)

#Esta última parte se hizo para probar distintos parámetros fácilmente
# best_global = (None, float("inf"), None)
# for tenure in [8, 10, 12, 14, 16, 18]:
#     for Tmax in [1000, 3000, 8000]:
#         sol, z = tabu_search_expiracion(sol_voraz, Tmax=Tmax, TabuTenure=tenure)
#         if z < best_global[1]:
#             best_global = (sol, z, (Tmax, tenure))
#         print("Tmax", Tmax, "tenure", tenure, "->", z)

```

**Apartado de diseño**

En el diseño del algoritmo voraz, me sugerió la siguiente mejora: utilizar el solapamiento de actores como criterio de desempate cuando dos tomas producen el mismo incremento en el número de actores necesarios. Este criterio favorece agrupar tomas que comparten actores, lo que reduce la probabilidad de tener que traer nuevos actores al estudio en días adicionales.

En el algoritmo de búsqueda tabú también me sugirió representar cada movimiento mediante un identificador único formado por el par de tomas intercambiadas para evitar considerar como movimientos distintos intercambios equivalentes (por ejemplo, intercambiar la toma A con la B o la B con la A).
